# Custom Russian Tokenizer

## Описание проекта

Учебный проект по созданию собственного токенизатора для русского языка
на основе алгоритма BPE с использованием библиотеки Hugging Face Tokenizers.

Для обучения используется русская часть книжного корпуса
[OPUS Books](https://huggingface.co/datasets/Helsinki-NLP/opus_books),
языковая пара `en-ru`.

## Цель

Обучить токенизатор с нуля и исследовать, как он обрабатывает русскую
морфологию, составные слова, современные заимствования и редкие имена.

## Основные этапы

1. Исследование и очистка корпуса.
2. Настройка и обучение BPE-токенизатора.
3. Сравнение с готовым токенизатором `cointegrated/rubert-tiny2`.
4. Расчёт метрик: длина последовательностей, количество токенов на слово
   и наличие неизвестных токенов `[UNK]`.
5. Анализ удачных и неудачных разбиений и формулирование выводов.

## ШАГ #1 — Загрузка локального корпуса OPUS Books

### Что делаем

Подключаем скачанную англо-русскую часть корпуса OPUS Books
и загружаем её в объект `Dataset` библиотеки Hugging Face Datasets.

### Последовательность действий

1. Импортируем `Path` для работы с путями и `load_dataset` для чтения данных.
2. Указываем папку с корпусом внутри контейнера:
   `/workspace/data/opus_books/en-ru`.
3. Находим файлы с расширением `.parquet` и сортируем их пути.
4. Проверяем наличие файлов. Если файлы отсутствуют, останавливаем
   выполнение с понятным сообщением.
5. Читаем локальные Parquet-файлы и сохраняем датасет в переменную `corpus`.
6. Выводим количество записей и названия полей.

### Формат данных

Parquet — бинарный табличный формат, поддерживающий сжатие
и хранение данных по столбцам.

Каждая запись корпуса содержит:

- `id` — идентификатор записи;
- `translation` — структуру с английским (`en`) и русским (`ru`) текстами.

### Ожидаемый результат

Получаем датасет из **17 496 записей** с полями `id` и `translation`.

На этом шаге читаем локальные файлы: повторная загрузка корпуса
из интернета не требуется. Тексты пока не очищаем и не изменяем.
Русскую часть извлечём на следующем шаге.

```python
# Path позволяет работать с путями к папкам и файлам.
from pathlib import Path

# load_dataset загружает данные и создаёт объект Dataset.
from datasets import load_dataset


# Путь к папке с англо-русской частью корпуса внутри Docker-контейнера.
# Папка проекта на компьютере подключена к контейнеру как /workspace.
data_dir = Path("/workspace/data/opus_books/en-ru")

# Находим в этой папке все файлы с расширением .parquet.
# glob возвращает найденные пути, sorted собирает их в отсортированный список.
# Сортировка обеспечивает одинаковый порядок файлов при каждом запуске.
parquet_files = sorted(data_dir.glob("*.parquet"))

# Если список пуст, значит файлы не найдены.
# Останавливаем выполнение и выводим понятное сообщение.
if not parquet_files:
    raise FileNotFoundError(f"В папке {data_dir} нет Parquet-файлов")

# Читаем локальные файлы и создаём датасет.
corpus = load_dataset(
    # Указываем формат файлов: Apache Parquet.
    "parquet",

    # Передаём список путей к файлам.
    # str(path) преобразует каждый объект Path в строку.
    data_files=[str(path) for path in parquet_files],

    # Получаем часть данных под названием train.
    # Это НЕ делит корпус на обучающую и тестовую выборки:
    # здесь все переданные файлы относятся к одной части train.
    split="train",
)

# Показываем краткое описание датасета:
# названия полей и количество записей.
print(corpus)

# len(corpus) возвращает число записей в датасете.
# \n добавляет пустую строку перед сообщением.
# Формат : , без пробела между символами — то есть :, —
# выводит число с разделителями тысяч: 17,496.
print(f"\nКоличество записей: {len(corpus):,}")

# column_names возвращает список названий полей.
# В нашем корпусе ожидаются поля id и translation.
print(f"Поля датасета: {corpus.column_names}")

# Количество записей, которые хотим показать.
# min защищает от выхода за границы, если корпус меньше пяти записей.
examples_count = min(5, len(corpus))

# Последовательно получаем записи по индексам от 0 до examples_count - 1.
for index in range(examples_count):
    # Одна запись Dataset возвращается как Python-словарь.
    record = corpus[index]

    # В поле translation находится вложенный словарь переводов.
    translation = record["translation"]

    print(f"Запись #{index} | id: {record['id']}")
    print(f"EN: {translation['en']}")
    print(f"RU: {translation['ru']}")

    # Разделяем примеры горизонтальной линией.
    print("-" * 80)
```

## ШАГ #2 — Извлечение и первичная проверка русских текстов

### Что делаем

Берём из каждой записи только русский текст — поле `translation["ru"]`.
Английский перевод для обучения нашего токенизатора не используется.

Проверяем:

1. Есть ли записи без русского текста.
2. Есть ли пустые строки или строки только из пробелов.
3. Встречаются ли лишние пробелы, табуляция и переносы строк.
4. Есть ли невидимые символы и служебные строки с указанием источника.

Выводим несколько русских текстов и примеры возможных проблем.

### Зачем это нужно

Артефакты форматирования и служебные записи могут попасть в словарь
токенизатора. Перед очисткой нужно посмотреть на реальные примеры
и определить обоснованные правила обработки.

### Ожидаемый результат

Получаем статистику качества корпуса и примеры для выбора правил очистки.
Исходные данные на этом шаге не изменяем.

```python
# re — стандартная библиотека Python для поиска текстовых шаблонов.
import re

# Counter — словарь-счётчик: удобно увеличивать количество найденных случаев.
from collections import Counter


# Генератор последовательно возвращает русские тексты.
# Он не создаёт дополнительный список всех текстов в оперативной памяти.
def iter_russian_texts(dataset):
    for record in dataset:
        yield record["translation"].get("ru")


# Счётчики найденных признаков.
stats = Counter({
    "Всего записей": 0,
    "Отсутствующий или нестроковый ru": 0,
    "Пустые строки": 0,
    "Пробельные артефакты": 0,
    "Невидимые символы": 0,
    "Служебный префикс": 0,
})

# Сохраняем только несколько примеров каждого признака.
problem_examples = {}

# Отдельно сохраним первые пять непустых русских текстов.
russian_examples = []


for index, text in enumerate(iter_russian_texts(corpus)):
    stats["Всего записей"] += 1

    # Перед обработкой проверяем тип значения.
    if not isinstance(text, str):
        stats["Отсутствующий или нестроковый ru"] += 1
        continue

    # strip() убирает пробелы и переносы строк по краям.
    # Если после этого ничего не осталось — строка пустая.
    if not text.strip():
        stats["Пустые строки"] += 1
        continue

    if len(russian_examples) < 5:
        russian_examples.append(text)

    # Каждый признак проверяем независимо:
    # одна запись может попасть сразу в несколько категорий.
    checks = {
        "Пробельные артефакты": (
            text != text.strip()
            or bool(re.search(r"\s{2,}|[\t\r\n\u00a0]", text))
        ),

        # BOM, пробел нулевой ширины и мягкий перенос.
        "Невидимые символы": any(
            character in text
            for character in "\ufeff\u200b\u00ad"
        ),

        # Это кандидаты для просмотра, а не готовое правило удаления.
        "Служебный префикс": bool(
            re.match(
                r"^\s*(?:Source:|Translation:|Источник:)",
                text,
                flags=re.IGNORECASE,
            )
        ),
    }

    for issue, found in checks.items():
        if found:
            stats[issue] += 1

            # Для каждой категории сохраняем максимум три примера.
            examples = problem_examples.setdefault(issue, [])

            if len(examples) < 3:
                examples.append((index, text))


print("СТАТИСТИКА КОРПУСА")
for name, count in stats.items():
    print(f"{name}: {count:,}")


print("\nПЕРВЫЕ РУССКИЕ ТЕКСТЫ")
for number, text in enumerate(russian_examples, start=1):
    print(f"{number}. {text}")


print("\nПРИМЕРЫ ВОЗМОЖНЫХ ПРОБЛЕМ")
if not problem_examples:
    print("Проверяемые признаки не обнаружены.")

for issue, examples in problem_examples.items():
    print(f"\n{issue}:")

    for index, text in examples:
        # repr() помогает увидеть специальные символы:
        # например, перевод строки отображается как \n.
        print(f"Запись #{index}: {text!r}")
```

## ШАГ #3 — Предобработка текстов и подготовка батчевого генератора

### Что делаем

1. Создаём функцию базовой очистки текста.
2. Приводим Unicode к форме NFC: одинаковые буквы, записанные разными
   последовательностями Unicode, получают единое представление.
3. Удаляем BOM, пробел нулевой ширины и мягкий перенос.
4. Заменяем последовательности пробельных символов одним обычным пробелом
   и убираем пробелы по краям.
5. Создаём генератор, который выдаёт очищенные тексты батчами.

### Что сохраняем

Сохраняем регистр, букву `ё`, дефисы, пунктуацию, названия книг
и заголовки. Не применяем лемматизацию и удаление стоп-слов:
токенизатор должен обучаться на естественных словоформах.

### Зачем нужны батчи

Генератор не создаёт дополнительный список всего очищенного корпуса.
Он накапливает небольшую порцию текстов и передаёт её для обучения
через `train_from_iterator`.

При этом сам алгоритм обучения токенизатора всё равно использует память
для хранения частот и построения словаря.

### Ожидаемый результат

Получаем функцию очистки и генератор батчей, готовый для обучения BPE.
Исходный объект `corpus` не изменяется.

```python
# unicodedata — стандартная библиотека для работы с Unicode.
import unicodedata


def clean_text(text):
    """Возвращает очищенный текст или пустую строку."""

    # Защита от пропущенных или некорректных значений.
    if not isinstance(text, str):
        return ""

    # NFC объединяет составные Unicode-последовательности,
    # если для них существует единый символ.
    text = unicodedata.normalize("NFC", text)

    # Удаляем отдельные технические символы.
    # Обычные буквы, пунктуацию и дефисы сохраняем.
    for character in "\ufeff\u200b\u00ad":
        text = text.replace(character, "")

    # split() без аргументов разбивает по пробельным символам.
    # join() соединяет части одним обычным пробелом.
    # Одновременно убираются пробелы по краям.
    text = " ".join(text.split())

    return text


def iter_clean_batches(dataset, batch_size=512):
    """Выдаёт списки очищенных текстов размером до batch_size."""

    if batch_size <= 0:
        raise ValueError("Размер батча должен быть положительным")

    # Здесь накапливается только текущая порция текстов.
    batch = []

    # Используем генератор русских текстов из предыдущего шага.
    for text in iter_russian_texts(dataset):
        cleaned_text = clean_text(text)

        # Пропускаем строки, ставшие пустыми после очистки.
        if not cleaned_text:
            continue

        batch.append(cleaned_text)

        # Когда батч заполнен, передаём его вызывающему коду.
        if len(batch) == batch_size:
            yield batch

            # После продолжения генератора начинаем новый батч.
            batch = []

    # Последний батч может быть меньше заданного размера.
    # Его тоже нужно передать, чтобы не потерять остаток данных.
    if batch:
        yield batch
```

## ШАГ #4 — Настройка и обучение BPE-токенизатора

### Что делаем

Создаём токенизатор с пустой моделью BPE и обучаем его на очищенных
русских текстах корпуса OPUS Books.

BPE начинает с отдельных символов и последовательно объединяет
часто встречающиеся пары в более крупные фрагменты.

### Настройки

- Размер словаря: до 10 000 токенов, включая специальные.
- Минимальная частота пары для слияния: 2.
- Нормализация Unicode: NFC.
- Предварительное разбиение: по пробелам с изоляцией пунктуации.
- Регистр и буква `ё` сохраняются.

### Специальные токены

- `[PAD]` — заполнение последовательностей до одинаковой длины.
- `[UNK]` — представление неизвестного символа.
- `[CLS]` — служебный токен для представления последовательности.
- `[SEP]` — разделитель последовательностей.
- `[MASK]` — обозначение скрытого фрагмента при обучении языковых моделей.

Резервируем эти токены в словаре. Автоматическое добавление `[CLS]`,
`[SEP]` и выравнивание длины последовательностей пока не настраиваем.

### Ожидаемый результат

Получаем обученный токенизатор и проверяем его на нескольких словах.
BPE выделяет частотные фрагменты, которые не обязательно совпадают
с корнями, приставками и окончаниями.

```python
# Tokenizer объединяет модель и этапы обработки текста.
from tokenizers import Tokenizer

# Импортируем компоненты для настройки токенизатора.
from tokenizers import models, normalizers, pre_tokenizers, trainers


# Максимальный размер словаря, включая специальные токены.
VOCAB_SIZE = 10_000

# Минимальная частота пары, необходимая для её слияния.
MIN_FREQUENCY = 2

# Количество текстов, передаваемых за один запрос к генератору.
BATCH_SIZE = 512

# Порядок специальных токенов определяет их начальные идентификаторы.
SPECIAL_TOKENS = ["[PAD]", "[UNK]", "[CLS]", "[SEP]", "[MASK]"]


# Создаём пустую BPE-модель: словарь и правила слияния ещё не обучены.
# Указываем токен для символов, которые модель не сможет представить.
tokenizer = Tokenizer(
    models.BPE(unk_token="[UNK]")
)

# Включаем NFC внутри самого токенизатора.
# Это обеспечит такую же нормализацию новых текстов при кодировании.
tokenizer.normalizer = normalizers.NFC()

# Настраиваем два последовательных этапа предварительного разбиения.
tokenizer.pre_tokenizer = pre_tokenizers.Sequence([
    # Разделяем текст по пробельным символам.
    pre_tokenizers.WhitespaceSplit(),

    # Изолируем каждый знак пунктуации.
    # Например, "Привет, мир!" сначала станет:
    # "Привет", ",", "мир", "!".
    pre_tokenizers.Punctuation(behavior="isolated"),
])

# Trainer управляет обучением словаря и правил слияния BPE.
trainer = trainers.BpeTrainer(
    vocab_size=VOCAB_SIZE,
    min_frequency=MIN_FREQUENCY,
    special_tokens=SPECIAL_TOKENS,
    show_progress=True,
)

# Проверяем предварительное разбиение до обучения модели.
example = "Привет, мир!"

print(
    tokenizer.pre_tokenizer.pre_tokenize_str(example)
)
```

```python
from time import perf_counter


# Запоминаем время начала обучения.
started_at = perf_counter()

# Создаём генератор батчей для обучения.
# Каждый вызов iter_clean_batches начинает новый проход по корпусу.
training_batches = iter_clean_batches(
    corpus,
    batch_size=BATCH_SIZE,
)

# Последовательно передаём батчи для обучения одной общей модели.
tokenizer.train_from_iterator(
    training_batches,
    trainer=trainer,
)

elapsed_seconds = perf_counter() - started_at

print(f"Обучение завершено за {elapsed_seconds:.1f} сек.")
print(f"Фактический размер словаря: {tokenizer.get_vocab_size():,}")
```

```python
# Смотрим, какие идентификаторы получили специальные токены.
print("СПЕЦИАЛЬНЫЕ ТОКЕНЫ")

for token in SPECIAL_TOKENS:
    print(f"{token}: {tokenizer.token_to_id(token)}")


# Первые примеры для проверки обученной модели.
examples = [
    "Привет, мир!",
    "переосмысливаемый",
    "недопонятый",
    "сельскохозяйственный",
    "загуглить",
]

print("\nПРИМЕРЫ ТОКЕНИЗАЦИИ")

for text in examples:
    # encode возвращает объект Encoding с токенами и их идентификаторами.
    encoding = tokenizer.encode(text)

    print(f"\nТекст: {text}")
    print(f"Токены: {encoding.tokens}")
    print(f"Идентификаторы: {encoding.ids}")
```

## ШАГ #5 — Сохранение и проверка токенизатора

### Что делаем

1. Создаём папку `artifacts` для результатов обучения.
2. Сохраняем токенизатор в JSON.
3. Отдельно сохраняем словарь соответствий «токен → ID».
4. Загружаем токенизатор из JSON и сравниваем результаты с исходной моделью.

### Какие файлы получим

- `bpe_ru_10000.json` — модель BPE, словарь, правила слияния,
  нормализация и настройки предварительного разбиения.
- `vocab_ru_10000.json` — отдельный словарь для просмотра и анализа.

Для повторного использования токенизатора достаточно первого файла.
Отдельный файл словаря не содержит всех настроек и правил слияния.

### Ожидаемый результат

Загруженная модель выдаёт те же токены и ID, что и модель после обучения.
Повторный запуск ячейки перезаписывает файлы.

```python
# json нужен для отдельного сохранения словаря.
import json

from pathlib import Path
from tokenizers import Tokenizer


# Папка результатов внутри подключённой папки проекта.
# Файлы сохранятся и на компьютере после остановки контейнера.
artifacts_dir = Path("/workspace/artifacts")

# Создаём папку, если она ещё не существует.
artifacts_dir.mkdir(parents=True, exist_ok=True)

# Используем размер словаря из настроек обучения в названиях файлов.
model_path = artifacts_dir / f"bpe_ru_{VOCAB_SIZE}.json"
vocab_path = artifacts_dir / f"vocab_ru_{VOCAB_SIZE}.json"


# Сохраняем весь настроенный токенизатор.
tokenizer.save(str(model_path))


# Получаем словарь вида {"токен": ID}.
vocab = tokenizer.get_vocab()

# Сортируем записи по ID для удобного просмотра.
sorted_vocab = dict(
    sorted(vocab.items(), key=lambda item: item[1])
)

# ensure_ascii=False сохраняет кириллицу в читаемом виде.
# indent=2 добавляет отступы в JSON.
vocab_path.write_text(
    json.dumps(sorted_vocab, ensure_ascii=False, indent=2),
    encoding="utf-8",
)

print(f"Модель сохранена: {model_path}")
print(f"Словарь сохранён: {vocab_path}")
print(f"Количество токенов в словаре: {len(vocab):,}")
```

```python
# Отдельная ячейка для проверки
# Создаём новый объект из сохранённого файла.
# Повторное обучение здесь не требуется.
loaded_tokenizer = Tokenizer.from_file(str(model_path))

test_texts = [
    "Привет, мир!",
    "переосмысливаемый",
    "недопонятый",
    "сельскохозяйственный",
    "загуглить",
]

for text in test_texts:
    # Кодируем одинаковый текст исходной и загруженной моделью.
    original = tokenizer.encode(text)
    restored = loaded_tokenizer.encode(text)

    # Если результаты отличаются, assert остановит выполнение.
    assert original.tokens == restored.tokens, (
        f"Не совпали токены для текста: {text}"
    )
    assert original.ids == restored.ids, (
        f"Не совпали ID для текста: {text}"
    )

# Проверяем также сохранность всего словаря.
assert tokenizer.get_vocab() == loaded_tokenizer.get_vocab()

print("Проверка пройдена: словари, токены и ID совпадают.")

print("\nПервые 20 записей словаря:")
for token, token_id in list(sorted_vocab.items())[:20]:
    print(f"{token_id:5} → {token!r}")
```

### Причина появления `[UNK]`

В имени «Эйяфьядлайёкюдль» собственный BPE заменил букву `ё`
токеном `[UNK]`.

Проверка показала:

- отдельный символ `ё` отсутствует в словаре токенизатора;
- RuBERT tiny2 обрабатывает это имя без `[UNK]`.

Таким образом, проблема связана с покрытием алфавита обучающими
данными. Увеличение размера словаря само по себе не добавит символ,
которого нет в корпусе.

Возможное исправление — задать русский алфавит через параметр
`initial_alphabet` при обучении BPE. Это обеспечит представление
буквы `ё`, но не гарантирует компактного разбиения слов с ней.

## ШАГ #8 — Пословное сравнение и оценка эффективности

### Что делаем

1. Выделяем слова из тестовых предложений.
2. Строим таблицу: слово → токены BPE → токены RuBERT.
3. Рассчитываем среднее количество токенов на слово.
4. Считаем неизвестные токены `[UNK]` и выводим проблемные слова.

### Правила подсчёта

Словом считаем непрерывную последовательность букв Unicode.
Пунктуацию, цифры и подчёркивания исключаем; дефис разделяет слова.

Каждое слово кодируем отдельно без специальных токенов.
Повторяющиеся слова учитываем при расчёте среднего,
но в таблице показываем каждое уникальное написание один раз.

Среднее количество токенов на слово:

общее количество токенов слов / количество слов.

Для `[UNK]` отдельно считаем количество и долю среди всех токенов
полных предложений, включая пунктуацию.

### Ограничения

Меньшее количество токенов означает более компактное представление,
но не обязательно более точное выделение морфем.

`[UNK]` может искусственно сокращать последовательность,
поэтому длины необходимо рассматривать вместе с покрытием символов.

```python
import re
import pandas as pd
from IPython.display import display


# Выделяем последовательности букв Unicode.
# \w включает буквы, цифры и подчёркивание.
# Конструкция [^\W\d_] оставляет буквы.
word_pattern = re.compile(r"[^\W\d_]+", flags=re.UNICODE)

# Сохраняем все вхождения слов, включая повторения.
# Выборка маленькая, поэтому здесь обычный список допустим.
test_words = []

# Мы тестируем на десяти предложениях, которые вручную записали в test_sentences на ШАГЕ #6
for sentence in test_sentences:
    text = clean_text(sentence)
    test_words.extend(word_pattern.findall(text))


def compare_word(word):
    """Возвращает разбиение слова и признаки наличия UNK."""

    custom_encoding = loaded_tokenizer.encode(
        word,
        add_special_tokens=False,
    )

    baseline_ids = baseline_tokenizer.encode(
        word,
        add_special_tokens=False,
    )

    baseline_tokens = baseline_tokenizer.convert_ids_to_tokens(
        baseline_ids
    )

    return {
        "Слово": word,
        "Токены BPE": " | ".join(custom_encoding.tokens),
        "Токены RuBERT": " | ".join(baseline_tokens),
        "Количество BPE": len(custom_encoding.ids),
        "Количество RuBERT": len(baseline_ids),
        "UNK BPE": (
            loaded_tokenizer.token_to_id("[UNK]")
            in custom_encoding.ids
        ),
        "UNK RuBERT": (
            baseline_tokenizer.unk_token_id
            in baseline_ids
        ),
    }


# dict.fromkeys убирает повторения, сохраняя порядок появления.
unique_words = list(dict.fromkeys(test_words))

word_comparison = pd.DataFrame([
    compare_word(word)
    for word in unique_words
])

print(f"Всего вхождений слов: {len(test_words)}")
print(f"Уникальных написаний: {len(unique_words)}")

# Показываем все строки небольшой диагностической таблицы.
with pd.option_context(
    "display.max_rows", None,
    "display.max_colwidth", None,
):
    display(word_comparison)
```

```python
# Таблица содержит уникальные слова.
# Индекс позволяет быстро найти результат по самому слову.
word_lookup = word_comparison.set_index("Слово")

# Суммируем количество токенов для ВСЕХ вхождений слов.
# Если слово повторяется, его вклад учитывается повторно.
custom_word_tokens = sum(
    int(word_lookup.loc[word, "Количество BPE"])
    for word in test_words
)

baseline_word_tokens = sum(
    int(word_lookup.loc[word, "Количество RuBERT"])
    for word in test_words
)


# ID всех токенов полных предложений.
# Используем результаты, сохранённые на шаге #7.
custom_sentence_ids = [
    token_id
    for result in comparison_results
    for token_id in result["custom_ids"]
]

baseline_sentence_ids = [
    token_id
    for result in comparison_results
    for token_id in result["baseline_ids"]
]

custom_unk_id = loaded_tokenizer.token_to_id("[UNK]")
baseline_unk_id = baseline_tokenizer.unk_token_id

custom_unk_count = custom_sentence_ids.count(custom_unk_id)
baseline_unk_count = baseline_sentence_ids.count(baseline_unk_id)


metrics = pd.DataFrame([
    {
        "Модель": "Наш BPE",
        "Размер словаря": loaded_tokenizer.get_vocab_size(),
        "Токенов на слово": custom_word_tokens / len(test_words),
        "Токенов в предложениях": len(custom_sentence_ids),
        "Средняя длина предложения": (
            len(custom_sentence_ids) / len(comparison_results)
        ),
        "Количество UNK": custom_unk_count,
        "Доля UNK, %": (
            100 * custom_unk_count / len(custom_sentence_ids)
        ),
    },
    {
        "Модель": "RuBERT tiny2",
        "Размер словаря": len(baseline_tokenizer),
        "Токенов на слово": baseline_word_tokens / len(test_words),
        "Токенов в предложениях": len(baseline_sentence_ids),
        "Средняя длина предложения": (
            len(baseline_sentence_ids) / len(comparison_results)
        ),
        "Количество UNK": baseline_unk_count,
        "Доля UNK, %": (
            100 * baseline_unk_count / len(baseline_sentence_ids)
        ),
    },
])

display(metrics.round(3))


# Выводим слова, содержащие неизвестные токены хотя бы у одной модели.
unknown_words = word_comparison[
    word_comparison["UNK BPE"] | word_comparison["UNK RuBERT"]
]

print("\nСлова с неизвестными токенами:")

if unknown_words.empty:
    print("Не обнаружены.")
else:
    with pd.option_context("display.max_colwidth", None):
        display(unknown_words)
```

### Удачные и неудачные разбиения

| Исходное слово | Токены нашего BPE | Токены RuBERT tiny2 |
|---|---|---|
| Переосмысливаемый | Пере · ос · мысли · ваемый | Пер · ##ео · ##см · ##ыс · ##лива · ##емый |
| переобъяснить | пере · объяснить | перео · ##б · ##ъя · ##снить |
| подземный | под · зем · ный | подземный |
| подводный | под · вод · ный | подвод · ##ный |
| кооператив | ко · опе · ра · тив | кооператив |
| загуглить | за · гу · гли · ть | загу · ##гли · ##ть |
| кринж | кри · н · ж | кри · ##н · ##ж |
| мем | ме · м | мем |

Точки между фрагментами обозначают границы токенов.
Метка `##` у RuBERT обозначает продолжение слова.

К удачным примерам BPE относятся `пере · объяснить`,
`под · зем · ный` и `под · вод · ный`:
модель выделяет фрагменты, соответствующие приставкам
и другим значимым частям слов.

Однако разбиение не всегда согласуется с морфологией.
Например, слово «Высококвалифицированная» разбивается на девять токенов:

`Вы · со · ко · к · вали · фи · ци · рова · нная`.

Такое дробление увеличивает длину последовательности.
В словах «кооператив» и «мем» также наблюдается менее компактное
представление, чем у RuBERT.

### Обработка русской морфологии

BPE обнаруживает повторяющиеся фрагменты без словаря морфем
и без знания грамматических правил.

В результатах отдельно выделяются `пере` и `под`.
При этом распространённые окончания не обязательно становятся
отдельными токенами: «нового», «доброго» и «молодая»
представлены целиком, а «красивого» — как `краси · вого`.
В слове «Синий» получено разбиение `Си · ний`.

Следовательно, результаты показывают усвоение некоторых
повторяющихся частей слов, но не позволяют утверждать,
что модель систематически выделяет окончания `-ого`, `-ая`, `-ий`.
BPE является статистическим токенизатором,
а не морфологическим анализатором.

### Неизвестные символы и редкие имена

В имени «Эйяфьядлайёкюдль» собственная модель выдала 13 токенов,
включая один `[UNK]`. RuBERT выдал 9 токенов без `[UNK]`.

Проверка смещений показала, что неизвестным фрагментом стала
буква `ё` (`U+0451`). В очищенном обучающем корпусе нет её вхождений,
а в словаре BPE отсутствует соответствующий символ.

Таким образом, новое слово из знакомых символов обычно можно
представить подсловами, но отсутствие самого символа
приводит к потере информации.

Возможное исправление — задать полный русский алфавит
через `initial_alphabet` при обучении.
В рамках проведённого эксперимента это исправление не проверялось.

### Влияние корпуса и размера словаря

Книжный корпус обеспечивает примеры литературных словоформ,
но не гарантирует достаточного покрытия современных заимствований,
сленга и редких имён. Полученные разбиения согласуются
с этим ограничением, однако влияние домена отдельно не измерялось.

Увеличение объёма и разнообразия данных может улучшить покрытие
слов и символов. При этом простое увеличение количества похожих текстов
не гарантирует появления отсутствующей буквы или современной лексики.

Больший словарь позволяет хранить больше длинных фрагментов
и целых слов, что обычно уменьшает дробление.
Одновременно растут требования к хранению словаря и,
при использовании в нейросети, к размеру таблицы эмбеддингов.

В работе проверен только размер 10 000.
Поэтому влияние изменения размера словаря обсуждается теоретически,
а не подтверждается отдельным экспериментом.

### Общий вывод

Собственный BPE успешно обучен и способен представлять большинство
проверенных слов через выученные фрагменты.
Он выделяет некоторые характерные русские приставки и части слов,
но чрезмерно дробит ряд сложных слов и не покрывает букву `ё`.

Готовый токенизатор RuBERT tiny2 на выбранных предложениях
создаёт более компактные последовательности и обеспечивает
лучшее покрытие символов. Однако его словарь примерно в 8,4 раза больше;
также различаются обучающие данные и алгоритм токенизации.

Поэтому эксперимент сравнивает два готовых результата,
но не доказывает преимущество WordPiece над BPE.
Компактность также не равнозначна правильности морфемных границ.

Десять предложений позволяют обнаружить характерные ограничения,
но для обобщения результатов нужны независимые тексты разных областей.
Для отдельного изучения влияния размера словаря можно обучить
BPE на том же корпусе с 20 000 токенов и повторить оценку.
